## 2. The question: decision, action, cost of a wrong call

### Decision

Which content pages should be reviewed first?

### Who acts?

A content editor or SEO reviewer.

### What action?

Review, refresh, expand, monitor, or postpone a page based on its priority score.

### Cost of a wrong recommendation

If a healthy page is ranked too highly, reviewer time is wasted.

If a declining page is ranked too low, it may continue losing visibility before anyone notices.

### Why data or ML?

Simple rules can identify obvious cases, but page performance depends on many interacting signals such as impressions, CTR, freshness, content age, and engagement. A ranking model can combine these signals more effectively while still supporting human review.

### Task Type

Ranking / Scoring

### Output

A ranked list of content pages with refresh priority scores.

### Success Metric

Precision@K (for example Precision@20 or Precision@50), because reviewers only inspect the highest-ranked pages.
