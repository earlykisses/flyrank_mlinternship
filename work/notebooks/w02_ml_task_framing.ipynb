## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# Example only (illustrative)

df["refresh_target"] = (
    (df["ctr"] < df["ctr"].median()) &
    (df["impressions_90d"] > df["impressions_90d"].median())
).astype(int)

df[["ctr","impressions_90d","refresh_target"]].head()
```

## Why ML beats a fixed rule

A fixed rule such as:

"If CTR < 3% then refresh"

does not capture the interaction between multiple signals.

Whether a page should be refreshed depends on combinations of search volume, impressions, ranking position, historical traffic, and content characteristics.

Machine learning can learn these complex relationships and produce a ranked list of opportunities rather than relying on a single threshold.
